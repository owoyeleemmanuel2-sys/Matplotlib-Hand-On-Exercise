## Ex 1: Retail Sales Trend Analysis

```python
# Importing matplotlib as plt 
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Creating Dataset
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [120000, 145000, 138000, 165000, 190000, 215000]

plt.plot(months, sales)
plt.xlabel("Months")
plt.ylabel("Sales (#)")
plt.title("Monthly Sales Trend")
plt.plot(months, sales, "k--", marker="o")
plt.show()
```

### Bonus Observations:
- Sales grew from 120k to 215k - 79% growth, but dropped in March (145k -> 138k).Strongest growth Apr-Jun, suggests seasonal demand or successful marketing.
#### Why line chart? 
- Because it shows change over continuous time. Best for trend.
#### Reflection:
- Line chart is appropriate because time is ordered and we want to see trend. I would request marketing spend, promotions, holidays, stock availability before explaining why sales increased/decreased.

## EX 2: Customer Acquisition Comparison

```python
# Creating Dataset
months = [1, 2, 3, 4, 5, 6]

campaign_a = [50, 65, 72, 90, 110, 125]

campaign_b = [45, 60, 68, 75, 95, 105]

plt.title("Customer Acquisition Comparison")
plt.xlabel("Months")
plt.ylabel("Customers Chart")
plt.plot(campaign_a, "k--", campaign_b, "g--", marker="o")
plt.axis([1, 6, 40, 150])
plt.show()
```

### Observations:
- Campaign A always higher. Gap widens from Month 3 (4 customers) to Month 6 (20 customers).
### Reflection:
- Chart shows A acquired more, but not enough to invest. Need: Cost per acquisition, budget spent, customer lifetime value, conversion quality, channel type.

## EX 3: Healthcare Performance Dashboard

```python
# Dataset
months = [1, 2, 3, 4, 5, 6]
patients = [420, 450, 480, 510, 495, 540]
waiting_time = [65, 62, 58, 55, 60, 49]
satisfaction = [72, 74, 76, 78, 77, 83]

# Create figure with 3 vertically arranged subplots
fig, axes = plt.subplots(3, 1, figsize=(8, 12))

# 1. Patients
axes[0].plot(months, patients, marker='o', color='blue')
axes[0].set_title('Number of Patients Treated per Month')
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Patients (count)')

# 2. Waiting Time
axes[1].plot(months, waiting_time, marker='s', color='red')
axes[1].set_title('Average Waiting Time per Month')
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Waiting Time (minutes)')

# 3. Satisfaction
axes[2].plot(months, satisfaction, marker='^', color='green')
axes[2].set_title('Patient Satisfaction Score per Month')
axes[2].set_xlabel('Month')
axes[2].set_ylabel('Satisfaction (0-100)')

# Improve spacing
plt.tight_layout()

# Show
plt.show()
```

#### Why not plot 3 lines on same y-axis?
- Because they have different units and scales. Patients = 420-540 counts, waiting = 49-65 minutes, satisfaction = 72-83 score. If you put them on one y-axis (0-540), the waiting and satisfaction lines will look flat at the bottom and you can't see changes. You must use separate subplots or normalize scales. This avoids misleading visuals.
#### What happened to patient volume?
- Increased overall from 420 to 540, with steady growth Months 1-4, small drop in Month 5, then peak in Month 6.2. How did average waiting time change?
Generally decreased from 65 to 49 mins. Trend was downward Months 1-4, up in Month 5 to 60 mins, then sharpest improvement in Month 6.3. 
#### What happened to satisfaction in final month?
- Jumped significantly to 83 in Month 6 - the highest score, up from 77 in Month 5.4. 
Does shorter waiting time cause higher satisfaction?
No, patterns show correlation, not causation. While both improved together, we cannot say shorter wait caused satisfaction. Other factors could be involved (more doctors, better care, facility improvements). Correlation does not equal causation.5. 
#### What additional data would help?
- Number of staff/doctors per month
- Type of cases / patient severity
- Staff-to-patient ratio
- Feedback comments on why patients are satisfied
- Time of day / department data
- Number of complaints

## Mini Project: 
### Business Performance Visualization

### Project title: Visualizing Business Performance with Matplotlib

```python
# creating prject dataset

months = [1, 2, 3, 4, 5, 6]

revenue = [150000, 165000, 158000, 190000, 215000, 240000]

customers = [80, 92, 88, 110, 125, 140]

product_a = [45, 50, 48, 60, 68, 75]

product_b = [35, 40, 42, 45, 52, 58]

# Create figure with 3 vertically arranged subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# 1. Monthly Revenue Trend
axes[0,0].plot(months, revenue, marker= "o", color = "red")
axes[0,0].set_title("Monthly Revenue")
axes[0,0].set_xlabel("Month")
axes[0,0].set_ylabel("Revenue (#)")

# 2. Customer growth
axes[0,1].plot(months, customers, marker= "s", color = "blue")
axes[0,1].set_title("Customer Growth")
axes[0,1].set_xlabel("Month")
axes[0,1].set_ylabel("Customers")

# 3. Product performance comparison

axes[1,0].plot(months, product_a, marker = "o", color = "black", label= 'Product A', linewidth=2)
axes[1,0].plot(months, product_b,marker = "s", color = "green", label= 'Product B',linestyle='--', linewidth=2)
axes[1,0].set_title("Product Performance Comparison")
axes[1,0].set_xlabel("Month")
axes[1,0].set_ylabel("Unit Sold")
axes[1,0].legend()
axes[1,1].axis('off') # leave 4th empty
plt.tight_layout()
# Show
plt.show()
```

### Business Report (3 observations):
- Overall Revenue Up: From 150k to 240k (+60%). Only decline Month 3 (165k -> 158k) same month customers dropped.
- Customer trend matches revenue: Customers 80->140, correlates with revenue growth. Month 3 dip in both.
- Product A leads: Product A higher every month, final month 75 vs 58 units. Both growing but A is driver.
#### What additional info needed?
- Marketing campaigns, pricing changes, seasonality, product stock and customer feedback.
